EXP 14- DATA BINDING AND DATA FORMATTING USING PYTHON

```python
import pandas as pd
import numpy as np
#conversion of numeric data to categorical data such as low medium high etc etc for different data.

data = {
'Product':['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
'Price':[55000,20000,30000,2000,8000,45000],
'Units_Sold':[25,60,35,80,50,15]
}

df = pd.DataFrame(data)
print(df)

```

```python
#Data Binning
bins1=[0,10000,30000,60000]
labels1=['Cheap','Medium','Expensive']
df['Price_Category'] = pd.cut(df['Price'],bins=bins1, labels=labels1)

#pd.cut converts continous data tocategorical data

print(df)
```

```python
#Binning Units Sold
bins2=[0,15,30,60,100]
labels2=['Very Low Sales','Low Sales','Medium sales','High sales']
df['Sales_Category'] = pd.cut(df['Units_Sold'],bins=bins2, labels=labels2)
print(df)
```

```python
#Data Formatting
#Convert Data Type

print(df.dtypes)

#df['Units_Sold] = df['Units_Sold'].astype(float) #conversion of unit_sold from int to float
#print(df.dtypes)
```

```python
print(df.dtypes)

df['Units_Sold'] = df['Units_Sold'].astype(float)

#conversion of unit_sold from int to float shown here

print(df.dtypes)
```

```python
#Convert text to Uppercase
df['Product']= df['Product'].str.upper()
print(df)
```

```python
#Rounding Values
df['Price']=df['Price'].round(2)
df
```

```python
#Additional Operations
#Sort Data
sorted_df= df.sort_values(by='Price')    #(if you want data in same table then use inplace=True after Price)
print(sorted_df)
```

```python
sorted_df= df.sort_values(by='Price',ascending=False) #Converts in descending order
print(sorted_df)
```

```python
#Display Unique Values
print(df['Price_Category'].unique())
```

```python
#Dataset: Food Delivery Orders
#Do all above function here also

import pandas as pd
import numpy as np

data = {
'Order_ID':['O1','O2','O3','O4','O5','O6','O7','O8'],
'Order_Value':[250,450,120,600,300,150,500,220],
'Delivery_Time':[30,45,25,60,35,20,50,28],
'Distance_km':[5,8,3,10,6,2,9,4]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Binning Order Value
bins=[0,200,400,700]
labels=['Low Order','Medium Order','High Order']
df['Order_Category']=pd.cut(df['Order_Value'],bins=bins,labels=labels)
print(df)
```

```python
#Count of categories
print(df['Order_Category'].value_counts())
df
```

```python
#Binning dilivery time
bins3=[0,25,45,65]
labels3=['Fast','Medium','Slow']
df['Delivery_Category']=pd.cut(df['Delivery_Time'],bins=bins3,labels=labels3)

print(df)
```

```python
#Binning distance
bins4=[0,3,7,11]
labels4=['near','medium','far']
df['Distance_Category']=pd.cut(df['Distance_km'],bins=bins4,labels=labels4)

print(df)
```

```python
#counting the value of categories
print(df['Order_Category'].value_counts())
print()
print(df['Delivery_Category'].value_counts())
print()
print(df['Distance_Category'].value_counts())
```

```python
#Data formatng
# Convert Data type

print(df.dtypes)

df['Delivery_Time']=df['Delivery_Time'].astype(float)
print(df.dtypes)
print()
print("updates table:")
print(df)
```
